```python
from functools import reduce

def odd(x):
    if x % 2 != 0:
        return x
def double(x):
    return x * 2
def sub(x, y):
    return x - y

num = [2,8,9,3,7,8]

odd_number = filter(odd, num)

double_num = map(double, odd_number)
result = reduce(sub, double_num)

print(result)
```

```python
def divisible(x):
    if x % 5 == 0:
        return x

num = [10,12,15,20,23,25,30]

result = list(filter(divisible, num))

total = sum(result)

count = 0
for x in result:
    count = count + 1

average = total / count

print("Numbers:", result)
print("Average:", average)
```

```python
from functools import reduce

def div3(x):
    if x % 3 == 0:
        return x

def positive(x):
    return abs(x)

def square(x):
    return x * x

def add(x, y):
    return x + y

num = [-12,5,-7,18,0,21,-3,10]

result = filter(div3, num)
result = map(positive, result)
result = map(square, result)
result = reduce(add, result)

print(result)
```

```python
from functools import reduce

def min_max(x, y):
    small = x[0]
    large = x[1]

    if y < small:
        small = y

    if y > large:
        large = y

    return (small, large)

num = [34,11,78,5,46,92,17]

result = reduce(min_max, num, (num[0], num[0]))

print("Smallest:", result[0])
print("Largest:", result[1])
```

```python
from functools import reduce

def difference(x, y):
    if y > x[1]:
        x = (x[0], y)

    if y < x[0]:
        x = (y, x[1])

    return x

num = [45,12,89,23,67,4,91]

result = reduce(difference, num, (num[0], num[0]))

print(result[1] - result[0])
```

```python
from functools import reduce

def count_values(result, x):
    if x > 0:
        result[0] = result[0] + 1
    elif x < 0:
        result[1] = result[1] + 1
    else:
        result[2] = result[2] + 1

    return result

num = [-5,0,12,-8,7,0,-2,15]

result = reduce(count_values, num, [0, 0, 0])

print("Positive number :", result[0])
print("Negative number:", result[1])
print("Zero:", result[2])
```

```python


from functools import reduce

def calculate(result, x):
    result[0] = result[0] + x
    result[1] = result[1] + 1
    return result

num = [18, 25, 32,47,58]

result = reduce(calculate, num, [0, 0])
average = result[0] / result[1]
print(average)
```

```python
from functools import reduce

def find_second(result, x):
    largest = result[0]
    second = result[1]
    if x > largest:
        second = largest
        largest = x

    elif x > second and x != largest:
        second = x

    return (largest, second)

num = [45,72,18,72,63,45,59]

result = reduce(find_second, num, [float('-inf'), float('-inf')])

print(result[1])
```

```python
from functools import reduce

def cumulative(result, x):
    
    total = result[-1] + x
    
    result.append(total)

    return result

num = [5,10,15,20,25]

result = reduce(cumulative, num, [0])

result = result[1:]

print(result)
```

```python
from functools import reduce

def remove_duplicate(result, x):
    if x not in result:
        result.append(x)

    return result

num = [4,7,4,2,7,9,2,5]

result = reduce(remove_duplicate, num, [])

print(result)
```
